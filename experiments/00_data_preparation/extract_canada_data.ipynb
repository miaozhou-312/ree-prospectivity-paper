-  This code efficiently extracts Canadian geochemical data from a massive global database by processing the CSV file in manageable chunks, standardizing the country_id field to ensure accurate filtering, and saving the resulting subset to a new CSV file.

```python
import pandas as pd

# 1. Define file paths for the giant global database and the localized output subset
file_path = r"C:\Users\mt125\OneDrive - Imperial College London\Desktop\Getech_REE_MinDBs_RockGeochemt_20260528\GlobalWholeRockGeochemicalDB\3359791\complete.csv"
output_path = r"C:\Users\mt125\OneDrive - Imperial College London\Desktop\Getech_REE_MinDBs_RockGeochemt_20260528\GlobalWholeRockGeochemicalDB\3359791\canada_geochem_subset.csv"

print("Scanning the global database in chunks (Filtering for country_id == 32)...")
canada_chunks = []
chunk_count = 0

# 2. Process the massive file in chunks of 100,000 rows to prevent MemoryError.
#    Using encoding_errors='replace' safely handles non-standard characters from multi-source global literature.
for chunk in pd.read_csv(file_path, 
                        encoding='utf-8', 
                        encoding_errors='replace', 
                        low_memory=False, 
                        chunksize=100000):
    chunk_count += 1
    
    # 3. Standardize the country_id column by converting to string and stripping spaces.
    #    Crucial step: Removes the '.0' float suffix automatically generated by Pandas due to missing values (NaNs).
    clean_id = chunk['country_id'].astype(str).str.strip().str.replace('.0', '', regex=False)
    
    # 4. Filter and lock onto Canadian sample points (where country_id is precisely '32')
    filtered_chunk = chunk[clean_id == '32']
    
    # Append the filtered subset if it is not empty
    if not filtered_chunk.empty:
        canada_chunks.append(filtered_chunk)
        
    print(f"Processed {chunk_count * 100:.0f}k rows... Total Canadian samples collected so far: {sum(len(c) for c in canada_chunks)}")

# 5. Concatenate all processed fragments and save the isolated asset locally
if canada_chunks:
    canada_data = pd.concat(canada_chunks, ignore_index=True)
    canada_data.to_csv(output_path, index=False, encoding='utf-8')
    print(f"\n🎉 Success! Extraction complete. Final shape of Canadian dataset: {canada_data.shape}")
    print(f"💡 Isolated subset has been saved to:\n{output_path}")
else:
    print("\n❌ Warning: Scan finished but no data found. Please double check if '32' is the correct country_id.")
```
